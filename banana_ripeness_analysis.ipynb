# 🍌 Banana Ripeness Detection: Deep Learning & Image Classification
### Dual-Engine Pipeline: Convolutional Neural Network (CNN) + Computer Vision

This notebook provides the complete Machine Learning workflow:
1. Loading & downloading real banana image datasets into standard categories (`unripe`, `ripe`, `overripe`)
2. Fine-tuning a pre-trained **MobileNetV2** Convolutional Neural Network using PyTorch
3. Running inference directly on image pixel data
4. Combining with Computer Vision (HSV Spectral Segmentation) for explainable results

```python
import os
import torch
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

from dataset_setup import download_sample_dataset
from cnn_detector import DeepBananaClassifier
from ripeness_detector import BananaRipenessDetector

print("✅ Libraries imported. PyTorch version:", torch.__version__)
print("CUDA GPU Available:", torch.cuda.is_available())
```

## 1. Setting Up the Image Dataset
Downloads real banana image samples into `dataset/train/` and `dataset/val/` categorized by ripeness stage.

```python
download_sample_dataset()

# Verify classes
for category in ["unripe", "ripe", "overripe"]:
    files = os.listdir(f"dataset/train/{category}")
    print(f"Category [{category}]: {len(files)} training images")
```

## 2. Train / Fine-Tune MobileNetV2 on Image Data

```python
from train_cnn import train_banana_model

# Trains for 4 epochs on CPU or GPU
model_path = train_banana_model(data_dir="dataset", num_epochs=4)
```

## 3. Deep Learning Inference on Image Data
Pass any raw image into the neural network to get class probabilities and confidence.

```python
classifier = DeepBananaClassifier()
sample_image = "dataset/train/ripe/ripe_1.jpg"

prediction = classifier.predict(sample_image)
print(f"Predicted Ripeness Class: {prediction['predicted_class']}")
print(f"Model Confidence: {prediction['confidence']}%")
print(f"Softmax Probabilities: {prediction['probabilities']}")
```

## 4. Visualizing Results with Predictions

```python
img = Image.open(sample_image)
plt.figure(figsize=(7, 7))
plt.imshow(img)
plt.title(f"CNN Prediction: {prediction['predicted_class']} ({prediction['confidence']}% Confident)", fontsize=14)
plt.axis('off')
plt.show()
```
